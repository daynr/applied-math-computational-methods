```python
from IPython.display import display, Markdown
import numpy as np
import matplotlib.pyplot as plt
```

```python
display(Markdown(r"""
# Problem 2.1(1): Interior of a Sphere

**Given:**
\(0 < r < a\), with boundary condition \(V(a) = V_0\), and \(V(r)\) remains bounded as \(r \to 0\).

Because the problem is radially symmetric, we write \(V = V(r)\).
Laplace’s equation becomes

$$
\nabla^2 V=\frac{1}{r^2}\frac{d}{dr}\left(r^2\frac{dV}{dr}\right)=0.
$$

So we solve

$$
\frac{d}{dr}\left(r^2\frac{dV}{dr}\right)=0.
$$

Integrating once gives

$$
r^2\frac{dV}{dr}=C_1.
$$

Hence

$$
\frac{dV}{dr}=\frac{C_1}{r^2}.
$$

Integrating again,

$$
V(r)=-\frac{C_1}{r}+C_2.
$$

Thus the general radially symmetric solution is

$$
V(r)=A+\frac{B}{r}.
$$

To remain bounded as \(r\to 0\), the term \(B/r\) must vanish, so \(B=0\). Therefore

$$
V(r)=A.
$$

Applying the boundary condition \(V(a)=V_0\) gives

$$
A=V_0.
$$

So the solution is

$$
\boxed{V(r)=V_0.}
$$

The radial electric field is

$$
E_r(r)=-\frac{dV}{dr}.
$$

Since \(V(r)=V_0\) is constant,

$$
\frac{dV}{dr}=0,
$$

so

$$
\boxed{E_r(r)=0.}
$$

### Physical Interpretation

The potential is constant throughout the interior, so there is no potential difference inside the sphere.
Therefore the electric field is zero everywhere in the interior.
"""))

# Parameters
a = 2.0
V0 = 5.0

# Domain
r = np.linspace(0, a, 400)

# Solutions
V = V0 * np.ones_like(r)
Er = np.zeros_like(r)

# Plot V(r)
plt.figure(figsize=(7,4))
plt.plot(r, V, linewidth=3, label=r"$V(r)=V_0$")
plt.xlabel(r"$r$")
plt.ylabel(r"$V(r)$")
plt.title("Problem 2.1(1a): Interior Potential")
plt.ylim(V0 - 1, V0 + 1)   # makes the flat line easier to see
plt.grid(True)
plt.legend()
plt.show()

# Plot E_r(r)
plt.figure(figsize=(7,4))
plt.plot(r, Er, linewidth=3, label=r"$E_r(r)=0$")
plt.xlabel(r"$r$")
plt.ylabel(r"$E_r(r)$")
plt.title("Problem 2.1(1b): Interior Electric Field")
plt.ylim(-1, 1)            # makes the zero line visible
plt.axhline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.show()
```

```python
display(Markdown(r"""
# Problem 2.1(2): Exterior of a Sphere

**Given:**
\(r > a\), with boundary condition \(V(a)=V_0\), and

$$
V(r)\to 0 \quad \text{as } r\to\infty.
$$

Because the problem is radially symmetric, we write \(V=V(r)\).
Then Laplace's equation becomes

$$
\nabla^2 V = \frac{1}{r^2}\frac{d}{dr}\left(r^2\frac{dV}{dr}\right)=0.
$$

So we solve

$$
\frac{d}{dr}\left(r^2\frac{dV}{dr}\right)=0.
$$

Integrating once gives

$$
r^2\frac{dV}{dr}=C_1.
$$

Hence

$$
\frac{dV}{dr}=\frac{C_1}{r^2}.
$$

Integrating again,

$$
V(r)=-\frac{C_1}{r}+C_2.
$$

Thus the general radially symmetric solution is

$$
V(r)=A+\frac{B}{r}.
$$

Now use the far-field condition \(V(r)\to 0\) as \(r\to\infty\). Since \(\frac{B}{r}\to 0\), we must have

$$
A=0.
$$

So the solution becomes

$$
V(r)=\frac{B}{r}.
$$

Apply the boundary condition \(V(a)=V_0\):

$$
V(a)=\frac{B}{a}=V_0.
$$

Therefore,

$$
B=aV_0.
$$

So the potential is

$$
\boxed{V(r)=\frac{aV_0}{r}.}
$$

---

### Electric Field

The radial electric field is

$$
E_r(r)=-\frac{dV}{dr}.
$$

Differentiate:

$$
\frac{dV}{dr}=\frac{d}{dr}\left(\frac{aV_0}{r}\right)=-\frac{aV_0}{r^2}.
$$

Therefore,

$$
E_r(r)=-\left(-\frac{aV_0}{r^2}\right)=\frac{aV_0}{r^2}.
$$

So

$$
\boxed{E_r(r)=\frac{aV_0}{r^2}.}
$$

---

### Far-field behavior

As \(r\to\infty\),

$$
V(r)=\frac{aV_0}{r}\to 0
$$

and

$$
E_r(r)=\frac{aV_0}{r^2}\to 0.
$$

So the potential decays like \(1/r\), while the electric field decays faster, like \(1/r^2\).
"""))

# Parameters
a = 2.0
V0 = 5.0
R = 10.0   # large enough to show far-field behavior

# Domain
r = np.linspace(a, R, 500)

# Solutions
V = a * V0 / r
Er = a * V0 / r**2

# Plot V(r)
plt.figure(figsize=(7,4))
plt.plot(r, V, linewidth=3, label=r"$V(r)=\frac{aV_0}{r}$")
plt.xlabel(r"$r$")
plt.ylabel(r"$V(r)$")
plt.title("Problem 2.1(2a): Exterior Potential")
plt.grid(True)
plt.legend()
plt.show()

# Plot E_r(r)
plt.figure(figsize=(7,4))
plt.plot(r, Er, linewidth=3, label=r"$E_r(r)=\frac{aV_0}{r^2}$")
plt.xlabel(r"$r$")
plt.ylabel(r"$E_r(r)$")
plt.title("Problem 2.1(2b): Exterior Electric Field")
plt.grid(True)
plt.legend()
plt.show()

display(Markdown(r"""
## Part (c): Far-Field Behavior

We analyze how \(V(r)\) and \(E_r(r)\) behave as \(r \to \infty\).

From the solution:

$$
V(r) = \frac{aV_0}{r}, \quad E_r(r) = \frac{aV_0}{r^2}.
$$

So:

- \(V(r)\) decays like \(1/r\)
- \(E_r(r)\) decays like \(1/r^2\)

Below, we confirm this behavior numerically.
"""))

# Check large-r behavior numerically
r_large = np.array([10, 20, 50, 100])

V_large = a * V0 / r_large
Er_large = a * V0 / r_large**2

print("r values:", r_large)
print("V(r):", V_large)
print("E_r(r):", Er_large)
```

```python
display(Markdown(r"""
# Problem 2.2: Exterior with Angular Dependence

We solve Laplace's equation in the exterior region \(r>a\), subject to

$$
V(a,\theta)=\cos\theta,
\qquad
V(r,\theta)\to 0 \quad \text{as } r\to\infty.
$$

Because this is a 2D polar-coordinate problem, we solve Laplace's equation in polar form:

$$
\nabla^2 V = V_{rr}+\frac{1}{r}V_r+\frac{1}{r^2}V_{\theta\theta}=0.
$$

Since the boundary data is proportional to \(\cos\theta\), we use the \(n=1\) separated solution

$$
V(r,\theta)=\left(Ar+\frac{B}{r}\right)\cos\theta.
$$

The condition \(V(r,\theta)\to 0\) as \(r\to\infty\) forces \(A=0\), so

$$
V(r,\theta)=\frac{B}{r}\cos\theta.
$$

Apply the boundary condition at \(r=a\):

$$
V(a,\theta)=\frac{B}{a}\cos\theta=\cos\theta.
$$

Hence,

$$
\frac{B}{a}=1
\quad\Rightarrow\quad
B=a.
$$

Therefore,

$$
\boxed{V(r,\theta)=\frac{a}{r}\cos\theta.}
$$

## Cartesian form on the \(xy\)-plane

Using

$$
r=\sqrt{x^2+y^2},
\qquad
\cos\theta=\frac{x}{r},
$$

we get

$$
V(x,y)=\frac{a}{r}\cdot\frac{x}{r}
=\frac{ax}{x^2+y^2}.
$$

So the Cartesian form is

$$
\boxed{V(x,y)=\frac{ax}{x^2+y^2}}, \qquad x^2+y^2>a^2.
$$

## Electric field

The electric field is

$$
\mathbf{E}=-\nabla V.
$$

For

$$
V(x,y)=\frac{ax}{x^2+y^2},
$$

we compute

$$
\frac{\partial V}{\partial x}
=
a\frac{y^2-x^2}{(x^2+y^2)^2},
\qquad
\frac{\partial V}{\partial y}
=
-\frac{2axy}{(x^2+y^2)^2}.
$$

Therefore,

$$
\boxed{
E_x = -\frac{\partial V}{\partial x}
= a\frac{x^2-y^2}{(x^2+y^2)^2},
\qquad
E_y = -\frac{\partial V}{\partial y}
= \frac{2axy}{(x^2+y^2)^2}.
}
$$

## Interpretation

- \(V(x,y)>0\) where \(x>0\)
- \(V(x,y)<0\) where \(x<0\)
- \(V(x,y)=0\) along the \(y\)-axis
- The field changes direction across the domain and is strongest near the boundary \(r=a\)
- Both the potential and field weaken as \(r\to\infty\)
"""))

# Parameters
a = 1.5
L = 5.0

# Grid
n = 400
x = np.linspace(-L, L, n)
y = np.linspace(-L, L, n)
X, Y = np.meshgrid(x, y)
R2 = X**2 + Y**2
R = np.sqrt(R2)

# Mask interior of circle
mask = R <= a

# Potential
V = np.full_like(X, np.nan, dtype=float)
V[~mask] = a * X[~mask] / R2[~mask]

# Electric field
Ex = np.full_like(X, np.nan, dtype=float)
Ey = np.full_like(X, np.nan, dtype=float)

Ex[~mask] = a * (X[~mask]**2 - Y[~mask]**2) / (R2[~mask]**2)
Ey[~mask] = 2 * a * X[~mask] * Y[~mask] / (R2[~mask]**2)

# Boundary circle
t = np.linspace(0, 2*np.pi, 400)

# Contour plot
plt.figure(figsize=(7,6))
cont = plt.contourf(X, Y, V, levels=30)
plt.colorbar(cont, label=r"$V(x,y)$")
plt.plot(a*np.cos(t), a*np.sin(t), linewidth=2)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Problem 2.2(2): Contour Plot of the Potential")
plt.axis("equal")
plt.show()

# Quiver plot on coarser grid
nq = 25
xq = np.linspace(-L, L, nq)
yq = np.linspace(-L, L, nq)
Xq, Yq = np.meshgrid(xq, yq)
R2q = Xq**2 + Yq**2
Rq = np.sqrt(R2q)
maskq = Rq <= a

Exq = np.full_like(Xq, np.nan, dtype=float)
Eyq = np.full_like(Xq, np.nan, dtype=float)

Exq[~maskq] = a * (Xq[~maskq]**2 - Yq[~maskq]**2) / (R2q[~maskq]**2)
Eyq[~maskq] = 2 * a * Xq[~maskq] * Yq[~maskq] / (R2q[~maskq]**2)

# normalize arrows so direction is easier to see
mag = np.sqrt(Exq**2 + Eyq**2)
Exn = Exq / mag
Eyn = Eyq / mag

plt.figure(figsize=(7,6))
plt.quiver(Xq, Yq, Exn, Eyn)
plt.plot(a*np.cos(t), a*np.sin(t), linewidth=2)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Problem 2.2(3): Quiver Plot of the Electric Field")
plt.axis("equal")
plt.show()
```

```python
display(Markdown(r"""
# Problem 2.3: Exterior with Full Angular Dependence

We solve Laplace's equation in the exterior region \(r>a\), subject to

$$
V(a,\theta,\phi)=\sin\theta\cos\phi,
\qquad
V(r,\theta,\phi)\to 0 \quad \text{as } r\to\infty.
$$

Because the boundary data is \(\sin\theta\cos\phi\), this is an \(l=1\) spherical harmonic mode.
The decaying exterior \(l=1\) solution has the form

$$
V(r,\theta,\phi)=\frac{B}{r^2}\sin\theta\cos\phi.
$$

Apply the boundary condition at \(r=a\):

$$
V(a,\theta,\phi)=\frac{B}{a^2}\sin\theta\cos\phi=\sin\theta\cos\phi.
$$

Hence,

$$
\frac{B}{a^2}=1
\quad\Rightarrow\quad
B=a^2.
$$

Therefore,

$$
\boxed{
V(r,\theta,\phi)=\frac{a^2}{r^2}\sin\theta\cos\phi
}.
$$

---

## Cartesian form

Using

$$
x=r\sin\theta\cos\phi,
$$

we get

$$
\sin\theta\cos\phi=\frac{x}{r}.
$$

So

$$
V(r,\theta,\phi)=\frac{a^2}{r^2}\cdot \frac{x}{r}
=\frac{a^2x}{r^3}.
$$

Since \(r=\sqrt{x^2+y^2+z^2}\), the Cartesian form is

$$
\boxed{
V(x,y,z)=\frac{a^2x}{(x^2+y^2+z^2)^{3/2}}
},
\qquad x^2+y^2+z^2>a^2.
$$

---

## Electric field

The electric field is

$$
\mathbf{E}=-\nabla V.
$$

For

$$
V(x,y,z)=a^2x(x^2+y^2+z^2)^{-3/2},
$$

we compute

$$
\frac{\partial V}{\partial x}
=
a^2\left[(x^2+y^2+z^2)^{-3/2}-3x^2(x^2+y^2+z^2)^{-5/2}\right]
=
a^2\frac{y^2+z^2-2x^2}{(x^2+y^2+z^2)^{5/2}},
$$

$$
\frac{\partial V}{\partial y}
=
-3a^2xy(x^2+y^2+z^2)^{-5/2},
$$

$$
\frac{\partial V}{\partial z}
=
-3a^2xz(x^2+y^2+z^2)^{-5/2}.
$$

Therefore,

$$
\boxed{
E_x=a^2\frac{2x^2-y^2-z^2}{(x^2+y^2+z^2)^{5/2}},
\quad
E_y=3a^2\frac{xy}{(x^2+y^2+z^2)^{5/2}},
\quad
E_z=3a^2\frac{xz}{(x^2+y^2+z^2)^{5/2}}.
}
$$

---

## Interpretation

- The solution is **not axisymmetric**.
- In the axisymmetric case, the solution depends only on \(r\) and \(\theta\), so rotating around the symmetry axis does not change the potential.
- Here, the extra \(\phi\)-dependence means the potential changes as we rotate around the axis.
- The boundary data is positive on the hemisphere where \(x>0\), negative where \(x<0\), and zero on the plane \(x=0\).
- So this behaves like a dipole oriented along the \(x\)-direction.
"""))

# Parameters
a = 1.5

# Part 1: boundary data on the sphere
theta = np.linspace(0, np.pi, 200)
phi = np.linspace(0, 2*np.pi, 200)
TH, PH = np.meshgrid(theta, phi)

Xs = a * np.sin(TH) * np.cos(PH)
Ys = a * np.sin(TH) * np.sin(PH)
Zs = a * np.cos(TH)

V_boundary = np.sin(TH) * np.cos(PH)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(
    Xs, Ys, Zs,
    facecolors=plt.cm.viridis((V_boundary - V_boundary.min())/(V_boundary.max()-V_boundary.min())),
    rstride=1, cstride=1, linewidth=0, antialiased=False, shade=False
)

mappable = plt.cm.ScalarMappable(cmap='viridis')
mappable.set_array(V_boundary)
fig.colorbar(mappable, ax=ax, shrink=0.7, pad=0.1, label=r"$V(a,\theta,\phi)$")

ax.set_title("Problem 2.3(1): Boundary Data on the Sphere")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_box_aspect([1,1,1])
plt.show()

# Part 2: contour plot on the plane z = 0

L = 5.0
n = 400
x = np.linspace(-L, L, n)
y = np.linspace(-L, L, n)
X, Y = np.meshgrid(x, y)
Z = np.zeros_like(X)

R = np.sqrt(X**2 + Y**2 + Z**2)
mask = R <= a

Vslice = np.full_like(X, np.nan, dtype=float)
Vslice[~mask] = a**2 * X[~mask] / (R[~mask]**3)

t = np.linspace(0, 2*np.pi, 400)

plt.figure(figsize=(7,6))
cont = plt.contourf(X, Y, Vslice, levels=30)
plt.colorbar(cont, label=r"$V(x,y,0)$")
plt.plot(a*np.cos(t), a*np.sin(t), linewidth=2)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Problem 2.3(2): Contour Plot of the Exterior Potential on z=0")
plt.axis("equal")
plt.show()


# Part 3: quiver plot on the plane z = 0
nq = 25
xq = np.linspace(-L, L, nq)
yq = np.linspace(-L, L, nq)
Xq, Yq = np.meshgrid(xq, yq)
Zq = np.zeros_like(Xq)

Rq = np.sqrt(Xq**2 + Yq**2 + Zq**2)
maskq = Rq <= a

Ex = np.full_like(Xq, np.nan, dtype=float)
Ey = np.full_like(Xq, np.nan, dtype=float)

Ex[~maskq] = a**2 * (2*Xq[~maskq]**2 - Yq[~maskq]**2) / (Rq[~maskq]**5)
Ey[~maskq] = 3 * a**2 * Xq[~maskq] * Yq[~maskq] / (Rq[~maskq]**5)

mag = np.sqrt(Ex**2 + Ey**2)
Exn = Ex / mag
Eyn = Ey / mag

plt.figure(figsize=(7,6))
plt.quiver(Xq, Yq, Exn, Eyn)
plt.plot(a*np.cos(t), a*np.sin(t), linewidth=2)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Problem 2.3(3): Quiver Plot of the Electric Field on z=0")
plt.axis("equal")
plt.show()
```

```python
display(Markdown(r"""
# Problem 3.1: Visualizing $f(z)=z^n$ for $n=-1,-2,2,3$

We write

$$
z = re^{i\theta}.
$$

Then

$$
z^n = r^n e^{in\theta}.
$$

So:

$$
|z^n| = r^n,
\qquad
\arg(z^n)=n\theta \pmod{2\pi}.
$$

## Interpretation

- For \(n>0\), the origin is a **zero**.
- For \(n<0\), the origin is a **pole**.
- A zero of multiplicity \(m\) behaves like \(z^m\), so:
  - the magnitude goes to \(0\) like \(r^m\),
  - the phase winds around \(m\) times.
- Thus:
  - \(z^2\) has a zero of multiplicity 2,
  - \(z^3\) has a zero of multiplicity 3,
  - \(z^{-1}\) has a pole of order 1,
  - \(z^{-2}\) has a pole of order 2.
"""))

# grid
L = 2.0
N = 600
x = np.linspace(-L, L, N)
y = np.linspace(-L, L, N)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y

# avoid exact zero
eps = 1e-12
Zsafe = Z.copy()
Zsafe[np.abs(Zsafe) < eps] = eps + 0j

powers = [-1, -2, 2, 3]

for n in powers:
    F = Zsafe**n
    mag = np.abs(F)
    phase = np.angle(F)

    # use log scale for magnitude to make behavior near 0/infinity easier to see
    logmag = np.log10(1 + mag)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    # log-magnitude
    im1 = axes[0].imshow(
        logmag,
        extent=[-L, L, -L, L],
        origin="lower",
        aspect="equal"
    )
    axes[0].set_title(rf"$\log_{{10}}(1+|z^{{{n}}}|)$")
    axes[0].set_xlabel("Re(z)")
    axes[0].set_ylabel("Im(z)")
    plt.colorbar(im1, ax=axes[0])

    # phase
    im2 = axes[1].imshow(
        phase,
        extent=[-L, L, -L, L],
        origin="lower",
        cmap="hsv",
        aspect="equal",
        vmin=-np.pi,
        vmax=np.pi
    )
    axes[1].set_title(rf"$\arg(z^{{{n}}})$")
    axes[1].set_xlabel("Re(z)")
    axes[1].set_ylabel("Im(z)")
    plt.colorbar(im2, ax=axes[1], ticks=[-np.pi, 0, np.pi])

    plt.suptitle(rf"Visualizing $f(z)=z^{{{n}}}$", fontsize=14)
    plt.tight_layout()
    plt.show()
```

```python
display(Markdown(r"""
# Problem 3.2: Visualizing $f(z)=e^z$

Let

$$
z=x+iy.
$$

Then

$$
e^z=e^{x+iy}=e^x e^{iy}=e^x(\cos y+i\sin y).
$$

So the magnitude is

$$
|e^z|=e^x,
$$

and the argument is

$$
\arg(e^z)=y \pmod{2\pi}.
$$

## Interpretation

This shows that:

- the magnitude depends only on the real part \(x\),
- the angle depends only on the imaginary part \(y\).

So:

- moving to the **right** in the complex plane makes the magnitude grow exponentially,
- moving to the **left** makes the magnitude decay toward \(0\),
- moving **up or down** changes the angle periodically,
- the function never equals \(0\), since \(e^x>0\) for all real \(x\).

Thus, the graph reveals that \(e^z\) has:
- **no zeros**,
- **no poles**,
- periodic behavior in the imaginary direction,
- exponential growth in the real direction.
"""))

# Grid in the complex plane

Lx = 3
Ly = 3*np.pi
N = 600

x = np.linspace(-Lx, Lx, N)
y = np.linspace(-Ly, Ly, N)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y

F = np.exp(Z)

mag = np.abs(F)
phase = np.angle(F)

# use log-magnitude for better visualization
logmag = np.log10(1 + mag)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Magnitude plot

im1 = axes[0].imshow(
    logmag,
    extent=[-Lx, Lx, -Ly, Ly],
    origin="lower",
    aspect="auto"
)
axes[0].set_title(r"$\log_{10}(1+|e^z|)$")
axes[0].set_xlabel("Re(z)")
axes[0].set_ylabel("Im(z)")
plt.colorbar(im1, ax=axes[0])

# Phase plot

im2 = axes[1].imshow(
    phase,
    extent=[-Lx, Lx, -Ly, Ly],
    origin="lower",
    cmap="hsv",
    aspect="auto",
    vmin=-np.pi,
    vmax=np.pi
)
axes[1].set_title(r"$\arg(e^z)$")
axes[1].set_xlabel("Re(z)")
axes[1].set_ylabel("Im(z)")
plt.colorbar(im2, ax=axes[1], ticks=[-np.pi, 0, np.pi])

plt.suptitle(r"Problem 3.2: Visualization of $f(z)=e^z$", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
display(Markdown(r"""
# Problem 3.3: Visualizing $f(z)=e^{1/z}$

We study

$$
f(z)=e^{1/z}.
$$

Let

$$
z = x+iy.
$$

Then \(z=0\) is not in the domain, so we examine the behavior near \(z=0\).

## Series expansion near \(z=0\)

Using the exponential series,

$$
e^w=\sum_{n=0}^\infty \frac{w^n}{n!},
$$

with \(w=1/z\), we get

$$
e^{1/z} = \sum_{n=0}^\infty \frac{1}{n! z^n}
= 1 + \frac{1}{z} + \frac{1}{2!z^2} + \frac{1}{3!z^3} + \cdots
$$

This Laurent series has infinitely many negative-power terms.

Therefore, \(z=0\) is **not** a removable singularity and **not** a pole.

So \(z=0\) is an **essential singularity**.

## Why the plots show this

For a pole, the magnitude would blow up in a uniform way as \(z\to 0\).

But for \(e^{1/z}\), the behavior depends strongly on the direction of approach:

- along the positive real axis, \(1/z \to +\infty\), so \(e^{1/z}\to\infty\),
- along the negative real axis, \(1/z \to -\infty\), so \(e^{1/z}\to 0\),
- along purely imaginary directions, the magnitude stays bounded and the phase oscillates.

So near \(z=0\), the function behaves wildly and differently in different directions.
That directional instability is exactly what we expect from an essential singularity.
"""))

# Grid in the complex plane

L = 1.5
N = 700

x = np.linspace(-L, L, N)
y = np.linspace(-L, L, N)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y

# avoid z = 0
eps = 0.05
mask = np.abs(Z) < eps

F = np.full_like(Z, np.nan + 1j*np.nan, dtype=complex)
F[~mask] = np.exp(1 / Z[~mask])

mag = np.abs(F)
phase = np.angle(F)

# log magnitude for visibility
logmag = np.log10(1 + mag)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Magnitude plot

im1 = axes[0].imshow(
    logmag,
    extent=[-L, L, -L, L],
    origin="lower",
    aspect="equal"
)
axes[0].set_title(r"$\log_{10}(1+|e^{1/z}|)$")
axes[0].set_xlabel("Re(z)")
axes[0].set_ylabel("Im(z)")
plt.colorbar(im1, ax=axes[0])

# draw small excluded circle around z=0
circle = plt.Circle((0, 0), eps, fill=False)
axes[0].add_patch(circle)

# Phase plot

im2 = axes[1].imshow(
    phase,
    extent=[-L, L, -L, L],
    origin="lower",
    cmap="hsv",
    aspect="equal",
    vmin=-np.pi,
    vmax=np.pi
)
axes[1].set_title(r"$\arg(e^{1/z})$")
axes[1].set_xlabel("Re(z)")
axes[1].set_ylabel("Im(z)")
plt.colorbar(im2, ax=axes[1], ticks=[-np.pi, 0, np.pi])

circle2 = plt.Circle((0, 0), eps, fill=False)
axes[1].add_patch(circle2)

plt.suptitle(r"Problem 3.3: Visualization of $f(z)=e^{1/z}$", fontsize=14)
plt.tight_layout()
plt.show()
```
