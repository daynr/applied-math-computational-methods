```python
from IPython.display import display, Markdown

display(Markdown(r"""
# 1.1 Static Euler-Bernoulli Beam Equation

We consider the static Euler-Bernoulli beam equation

$$
EI\,w^{(4)}(x)=q(x), \qquad 0<x<L,
$$

where:
- $E$ is Young's modulus,
- $I$ is the second moment of area,
- $q(x)$ is the transverse load,
- $w(x)$ is the beam deflection.

We assume $EI$ is constant.
"""))
```

```python
display(Markdown(r"""
## Problem 1

Assume $q$ is a **uniform load**, so $q(x)=q_0$ where $q_0$ is constant.

The equation becomes

$$
EI\,w^{(4)}(x)=q_0
\quad\Longrightarrow\quad
w^{(4)}(x)=\frac{q_0}{EI}.
$$

We are given boundary conditions for a clamped-free beam:

$$
w(0)=0,\qquad w'(0)=0,\qquad w''(L)=0,\qquad w'''(L)=0.
$$

### Step 1: Integrate four times

Integrating once:

$$
w'''(x)=\frac{q_0}{EI}x+C_1.
$$

Integrating again:

$$
w''(x)=\frac{q_0}{2EI}x^2+C_1x+C_2.
$$

Integrating again:

$$
w'(x)=\frac{q_0}{6EI}x^3+\frac{C_1}{2}x^2+C_2x+C_3.
$$

Integrating again:

$$
w(x)=\frac{q_0}{24EI}x^4+\frac{C_1}{6}x^3+\frac{C_2}{2}x^2+C_3x+C_4.
$$

### Step 2: Apply the boundary conditions

From $w'''(L)=0$:

$$
0=\frac{q_0}{EI}L+C_1
\quad\Longrightarrow\quad
C_1=-\frac{q_0L}{EI}.
$$

From $w''(L)=0$:

$$
0=\frac{q_0}{2EI}L^2-\frac{q_0L}{EI}L+C_2
=\frac{q_0L^2}{2EI}-\frac{q_0L^2}{EI}+C_2.
$$

So,

$$
C_2=\frac{q_0L^2}{2EI}.
$$

From $w'(0)=0$:

$$
C_3=0.
$$

From $w(0)=0$:

$$
C_4=0.
$$

### Step 3: Final explicit solution

Thus,

$$
w(x)=\frac{q_0}{24EI}x^4-\frac{q_0L}{6EI}x^3+\frac{q_0L^2}{4EI}x^2.
$$

Factoring out $\frac{q_0}{24EI}$ gives

$$
\boxed{
w(x)=\frac{q_0}{24EI}\left(x^4-4Lx^3+6L^2x^2\right)
}
$$

This is the explicit deflection profile for the clamped-free beam under uniform load.
"""))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
q0 = 1.0
E = 1.0
I = 1.0
L = 1.0

# Grid
x = np.linspace(0, L, 500)

# Deflection
w = (q0/(24*E*I))*(x**4 - 4*L*x**3 + 6*L**2*x**2)

# Plot
plt.figure(figsize=(8,5))
plt.plot(x, w, linewidth=2)
plt.xlabel(r"$x$")
plt.ylabel(r"$w(x)$")
plt.title("Deflection Profile for a Clamped-Free Beam with Uniform Load")
plt.grid(True)
plt.show()
```

```python
display(Markdown(r"""
## Problem 2

We now find the **maximum deflection**, where it occurs, and how the parameters affect it.

From Problem 1,

$$
w(x)=\frac{q_0}{24EI}\left(x^4-4Lx^3+6L^2x^2\right).
$$

### Step 1: Differentiate

$$
w'(x)=\frac{q_0}{24EI}\left(4x^3-12Lx^2+12L^2x\right).
$$

Factor:

$$
w'(x)=\frac{q_0}{24EI}\cdot 4x(x^2-3Lx+3L^2).
$$

So,

$$
w'(x)=\frac{q_0}{6EI}\,x(x^2-3Lx+3L^2).
$$

To find critical points, solve $w'(x)=0$.

This gives

$$
x=0
$$

or

$$
x^2-3Lx+3L^2=0.
$$

The discriminant is

$$
(-3L)^2-4(1)(3L^2)=9L^2-12L^2=-3L^2<0,
$$

so the quadratic has no real roots.

Thus the only critical point is $x=0$.

Since the beam is clamped at $x=0$ and the deflection increases along the beam, the maximum deflection occurs at the free end:

$$
\boxed{x=L}
$$

### Step 2: Compute the maximum value

Evaluate $w(L)$:

$$
w(L)=\frac{q_0}{24EI}\left(L^4-4L\cdot L^3+6L^2\cdot L^2\right).
$$

Simplify:

$$
w(L)=\frac{q_0}{24EI}(L^4-4L^4+6L^4)
=\frac{q_0}{24EI}(3L^4).
$$

Hence,

$$
\boxed{
w_{\max}=w(L)=\frac{q_0L^4}{8EI}
}
$$

### Step 3: Parameter influence

The formula

$$
w_{\max}=\frac{q_0L^4}{8EI}
$$

shows:

- Deflection is **directly proportional** to the load $q_0$.
- Deflection grows like **$L^4$**, so beam length has a very strong effect.
- Deflection is **inversely proportional** to $E$.
  A larger Young's modulus means a stiffer beam and less bending.
- Deflection is **inversely proportional** to $I$.
  A larger second moment of area means greater bending resistance.

So longer beams bend much more, while stiffer materials and larger stiffness reduce deflection.
"""))
```

```python
display(Markdown(r"""
# 1.2 Dynamic Euler-Bernoulli Beam Equation

We consider the dynamic Euler-Bernoulli beam equation

$$
EI\,w_{xxxx} + \mu w_{tt} = q(x).
$$

For **free vibration**, there is no transverse load, so

$$
q(x)=0.
$$

Thus the PDE becomes

$$
EI\,w_{xxxx} + \mu w_{tt} = 0.
$$

We will take $\mu=1$ for simplicity, but keep it symbolic in the derivation.
""")) # I assume we take μ=1 for simplicity
```

```python
display(Markdown(r"""
## Problem 1

Use separation of variables **without boundary conditions**.

Assume a solution of the form

$$
w(x,t)=X(x)T(t).
$$

Substitute into the PDE:

$$
EI\,X''''(x)T(t)+\mu X(x)T''(t)=0.
$$

Divide by $X(x)T(t)$:

$$
EI\frac{X''''(x)}{X(x)}+\mu\frac{T''(t)}{T(t)}=0.
$$

Rearrange:

$$
EI\frac{X''''(x)}{X(x)}=-\mu\frac{T''(t)}{T(t)}=\lambda,
$$

where $\lambda$ is a separation constant.

So we obtain the two ODEs

$$
EI\,X''''=\lambda X,
$$

and

$$
\mu T''+\lambda T=0.
$$

Now define

$$
\beta^4=\frac{\lambda}{EI}.
$$

Then the spatial equation becomes

$$
X''''-\beta^4 X=0,
$$

and the time equation becomes

$$
T''+\frac{\lambda}{\mu}T=0.
$$

Since $\lambda=EI\beta^4$, the time equation is

$$
T''+\frac{EI}{\mu}\beta^4T=0.
$$

### Spatial solution

The general solution to

$$
X''''-\beta^4X=0
$$

is

$$
X(x)=c_1\cosh(\beta x)+c_2\sinh(\beta x)+c_3\cos(\beta x)+c_4\sin(\beta x).
$$

### Time solution

The general solution to

$$
T''+\omega^2T=0
$$

with

$$
\omega^2=\frac{EI}{\mu}\beta^4
$$

is

$$
T(t)=A\cos(\omega t)+B\sin(\omega t),
$$

where

$$
\omega=\beta^2\sqrt{\frac{EI}{\mu}}.
$$

### Final separated form

Thus the eigenfunction expression is

$$
\boxed{
w(x,t)=
\left[c_1\cosh(\beta x)+c_2\sinh(\beta x)+c_3\cos(\beta x)+c_4\sin(\beta x)\right]
\left[A\cos(\omega t)+B\sin(\omega t)\right]
}
$$

with

$$
\boxed{
\omega=\beta^2\sqrt{\frac{EI}{\mu}}
}.
$$
"""))
```

```python
display(Markdown(r"""
## Problem 2

Now apply the **simple support** boundary conditions

$$
w(0,t)=w(L,t)=0,
\qquad
w_{xx}(0,t)=w_{xx}(L,t)=0.
$$

Since $w(x,t)=X(x)T(t)$, these become boundary conditions on $X(x)$:

$$
X(0)=X(L)=0,
\qquad
X''(0)=X''(L)=0.
$$

From the general spatial solution

$$
X(x)=c_1\cosh(\beta x)+c_2\sinh(\beta x)+c_3\cos(\beta x)+c_4\sin(\beta x),
$$

the simply supported conditions lead to the standard sine eigenfunctions

$$
\boxed{
X_n(x)=\sin\left(\frac{n\pi x}{L}\right),
\qquad n=1,2,3,\dots
}
$$

with

$$
\boxed{
\beta_n=\frac{n\pi}{L}
}.
$$

Therefore the natural frequencies are

$$
\omega_n=\beta_n^2\sqrt{\frac{EI}{\mu}}
=\left(\frac{n\pi}{L}\right)^2\sqrt{\frac{EI}{\mu}}.
$$

So,

$$
\boxed{
\omega_n=\frac{n^2\pi^2}{L^2}\sqrt{\frac{EI}{\mu}}
}.
$$

### Discussion of the effect of flexural rigidity

Since

$$
\omega_n\propto \sqrt{EI},
$$

increasing the flexural rigidity makes the beam **stiffer** and increases all natural frequencies.

Also, the **mode shapes do not change** with $EI$ in this simply supported case.
Only the oscillation frequencies change.
"""))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
L = 1.0
E = 1.0
I = 1.0
mu = 1.0

# Eigenmodes
x = np.linspace(0, L, 600)

plt.figure(figsize=(8,5))
for n in range(1, 6):
    Xn = np.sin(n*np.pi*x/L)
    plt.plot(x, Xn, label=rf"$n={n}$")

plt.xlabel(r"$x$")
plt.ylabel(r"$X_n(x)$")
plt.title("First 5 Simply Supported Beam Eigenmodes")
plt.grid(True)
plt.legend()
plt.show()

# Mode numbers
n_vals = np.arange(1, 6)

# Natural frequencies for simply supported beam
omega_vals = [(n*np.pi/L)**2 * np.sqrt(E*I/mu) for n in n_vals]

# Plot
plt.figure(figsize=(8,5))
plt.plot(n_vals, omega_vals, marker='o', linewidth=2)

plt.xlabel("Mode number $n$")
plt.ylabel(r"Natural frequency $\omega_n$")
plt.title("Natural Frequencies for Simply Supported Beam")
plt.grid(True)

for n, omega in zip(n_vals, omega_vals):
    plt.text(n, omega, f"{omega:.2f}", ha='center', va='bottom')

plt.show()

# Frequencies
display(Markdown(r"""
## Natural Frequencies for the Simply Supported Beam
"""))

for n in range(1, 6):
    omega_n = (n*np.pi/L)**2 * np.sqrt(E*I/mu)
    print(f"n = {n}:  omega_{n} = {omega_n:.6f}")

# EI
EI_values = [0.5, 1.0, 2.0, 5.0]

print("\nFirst three natural frequencies for different EI values:\n")
for EI in EI_values:
    print(f"EI = {EI}")
    for n in range(1, 4):
        omega_n = (n*np.pi/L)**2 * np.sqrt(EI/mu)
        print(f"  n = {n}: omega_{n} = {omega_n:.6f}")
    print()
```

```python
display(Markdown(r"""
## Problem 3

Now assume a **classical cantilever beam**, with the same boundary conditions as in 1.1.1:

- clamped at $x=0$:
$$
X(0)=0,\qquad X'(0)=0,
$$

- free at $x=L$:
$$
X''(L)=0,\qquad X'''(L)=0.
$$

We begin with the general spatial solution

$$
X(x)=c_1\cosh(\beta x)+c_2\sinh(\beta x)+c_3\cos(\beta x)+c_4\sin(\beta x).
$$

Applying the clamped conditions at $x=0$ gives the standard reduced form

$$
X(x)=C\big(\cosh(\beta x)-\cos(\beta x)\big)
+
D\big(\sinh(\beta x)-\sin(\beta x)\big).
$$

Applying the free-end conditions at $x=L$ yields a homogeneous linear system in $C$ and $D$.
For a nontrivial solution, the determinant must vanish, which gives the cantilever eigenvalue equation

$$
\boxed{
\cosh(\beta L)\cos(\beta L)+1=0
}
$$

This transcendental equation determines the admissible values of $\beta$.

If we define

$$
\alpha=\beta L,
$$

then the eigenvalue equation becomes

$$
\boxed{
\cosh(\alpha)\cos(\alpha)+1=0
}
$$

and the natural frequencies are

$$
\boxed{
\omega_n=\beta_n^2\sqrt{\frac{EI}{\mu}}
=
\frac{\alpha_n^2}{L^2}\sqrt{\frac{EI}{\mu}}
}
$$

where $\alpha_n$ is the $n$th positive root of

$$
\cosh(\alpha)\cos(\alpha)+1=0.
$$
"""))
```

```python
import scipy.optimize as opt

def f(alpha):
    return np.cosh(alpha)*np.cos(alpha) + 1

# We search in intervals near the root locations
intervals = [
    (1.0, 2.5),
    (4.0, 5.5),
    (7.0, 8.5),
    (10.0, 11.5),
    (13.0, 14.5),
]

alphas = []
for a, b in intervals:
    root = opt.brentq(f, a, b)
    alphas.append(root)

print("First 5 positive roots of cosh(alpha) cos(alpha) + 1 = 0:\n")
for i, alpha in enumerate(alphas, start=1):
    print(f"alpha_{i} = {alpha:.9f}")

def cantilever_mode(x, alpha, L=1.0):
    beta = alpha / L
    sigma = (np.cosh(alpha) + np.cos(alpha)) / (np.sinh(alpha) + np.sin(alpha))
    return (np.cosh(beta*x) - np.cos(beta*x)
            - sigma*(np.sinh(beta*x) - np.sin(beta*x)))

x = np.linspace(0, L, 800)

plt.figure(figsize=(9,6))
for i, alpha in enumerate(alphas, start=1):
    mode = cantilever_mode(x, alpha, L)
    mode = mode / np.max(np.abs(mode))  # normalize for plotting
    plt.plot(x, mode, label=rf"$n={i}$")

plt.xlabel(r"$x$")
plt.ylabel(r"Normalized mode shape")
plt.title("First 5 Cantilever Beam Mode Shapes")
plt.grid(True)
plt.legend()
plt.show()
```

### Problem 3

```python
display(Markdown(r"""
# 1.3 Incorporating Forcing

We now consider the forced dynamic Euler-Bernoulli beam equation

$$
EI\,w_{xxxx}+\mu w_{tt}=q(x,t),
$$

with simply supported boundary conditions

$$
w(0,t)=w(L,t)=0,\qquad w_{xx}(0,t)=w_{xx}(L,t)=0,
$$

and the beam starts completely at rest:

$$
w(x,0)=0,\qquad w_t(x,0)=0.
$$
"""))
```

```python
display(Markdown(r"""
## Problem 1

We want a forcing that captures oscillatory motion. A natural choice is

$$
\boxed{
q(x,t)=F_0\sin\!\left(\frac{\pi x}{L}\right)\cos(\Omega t)
}
$$

where:
- $F_0$ is the forcing amplitude,
- $\Omega$ is the driving frequency.

This is a good choice because:
1. it is oscillatory in time,
2. it is compatible with the simply supported eigenfunctions,
3. it makes the PDE easy to solve using a single-mode ansatz.

### Step 1: Choose a matching solution form

Because the forcing has the spatial shape $\sin(\pi x/L)$, we try

$$
w(x,t)=y(t)\sin\!\left(\frac{\pi x}{L}\right).
$$

Then

$$
w_{tt}=y''(t)\sin\!\left(\frac{\pi x}{L}\right),
$$

and since

$$
\frac{d^4}{dx^4}\sin\!\left(\frac{\pi x}{L}\right)
=
\left(\frac{\pi}{L}\right)^4 \sin\!\left(\frac{\pi x}{L}\right),
$$

we have

$$
w_{xxxx}=y(t)\left(\frac{\pi}{L}\right)^4\sin\!\left(\frac{\pi x}{L}\right).
$$

Substitute into the PDE:

$$
EI\,y(t)\left(\frac{\pi}{L}\right)^4\sin\!\left(\frac{\pi x}{L}\right)
+\mu y''(t)\sin\!\left(\frac{\pi x}{L}\right)
=
F_0\sin\!\left(\frac{\pi x}{L}\right)\cos(\Omega t).
$$

Cancel the common factor $\sin(\pi x/L)$:

$$
\mu y''(t)+EI\left(\frac{\pi}{L}\right)^4 y(t)=F_0\cos(\Omega t).
$$

Define the first natural frequency

$$
\omega_1^2=\frac{EI}{\mu}\left(\frac{\pi}{L}\right)^4.
$$

Then the modal equation becomes

$$
y''+\omega_1^2 y=\frac{F_0}{\mu}\cos(\Omega t).
$$

### Step 2: Solve the ODE

We solve

$$
y''+\omega_1^2 y=\frac{F_0}{\mu}\cos(\Omega t),
\qquad y(0)=0,\quad y'(0)=0.
$$

For $\Omega\neq\omega_1$, a particular solution is

$$
y_p(t)=\frac{F_0/\mu}{\omega_1^2-\Omega^2}\cos(\Omega t).
$$

The homogeneous solution is

$$
y_h(t)=C_1\cos(\omega_1 t)+C_2\sin(\omega_1 t).
$$

So

$$
y(t)=C_1\cos(\omega_1 t)+C_2\sin(\omega_1 t)
+\frac{F_0/\mu}{\omega_1^2-\Omega^2}\cos(\Omega t).
$$

Apply the initial conditions.

From $y(0)=0$:

$$
C_1+\frac{F_0/\mu}{\omega_1^2-\Omega^2}=0
\quad\Longrightarrow\quad
C_1=-\frac{F_0/\mu}{\omega_1^2-\Omega^2}.
$$

From $y'(0)=0$:

$$
C_2=0.
$$

Thus

$$
\boxed{
y(t)=\frac{F_0/\mu}{\omega_1^2-\Omega^2}\big(\cos(\Omega t)-\cos(\omega_1 t)\big)
}
$$

and therefore

$$
\boxed{
w(x,t)=
\frac{F_0/\mu}{\omega_1^2-\Omega^2}
\big(\cos(\Omega t)-\cos(\omega_1 t)\big)
\sin\!\left(\frac{\pi x}{L}\right)
}
$$

where

$$
\boxed{
\omega_1=\left(\frac{\pi}{L}\right)^2\sqrt{\frac{EI}{\mu}}.
}
$$

### Resonance case

If $\Omega=\omega_1$, then resonance occurs and the solution takes the form

$$
\boxed{
y(t)=\frac{F_0}{2\mu\omega_1}\,t\sin(\omega_1 t)
}
$$

so

$$
\boxed{
w(x,t)=\frac{F_0}{2\mu\omega_1}\,t\sin(\omega_1 t)\sin\!\left(\frac{\pi x}{L}\right).
}
$$
"""))
```

```python
# Parameters
L = 1.0
E = 1.0
I = 1.0
mu = 1.0
F0 = 1.0
Omega = 20.0   # choose a non-resonant forcing frequency

omega1 = (np.pi / L)**2 * np.sqrt(E * I / mu)

def y_t(t):
    return (F0 / mu) / (omega1**2 - Omega**2) * (np.cos(Omega * t) - np.cos(omega1 * t))

# Space and time grids
x = np.linspace(0, L, 400)
times = [0.0, 0.05, 0.1, 0.2, 0.3]

plt.figure(figsize=(9, 6))
for t in times:
    w = y_t(t) * np.sin(np.pi * x / L)
    plt.plot(x, w, label=rf"$t={t}$")

plt.xlabel(r"$x$")
plt.ylabel(r"$w(x,t)$")
plt.title("Forced Beam Deflection Profiles for Several Times")
plt.grid(True)
plt.legend()
plt.show()
```

```python
display(Markdown(r"""
## Problem 2

To model Brownian or stochastic motion, we can add a random forcing term to the beam equation.

A simple formulation is

$$
\boxed{
EI\,w_{xxxx}+\mu w_{tt}= \sigma\,\eta(x,t)
}
$$

where:
- $\sigma>0$ controls the noise strength,
- $\eta(x,t)$ is a random process representing stochastic forcing.

### A practical modal-compatible choice

To remain consistent with simply supported boundary conditions, we may choose

$$
\boxed{
q(x,t)=\sigma \sin\!\left(\frac{\pi x}{L}\right)\,\xi(t)
}
$$

where $\xi(t)$ is a stochastic process.

Then using the single-mode ansatz

$$
w(x,t)=y(t)\sin\!\left(\frac{\pi x}{L}\right),
$$

the PDE reduces to the stochastic ODE

$$
\boxed{
y''+\omega_1^2 y=\frac{\sigma}{\mu}\,\xi(t).
}
$$

### Brownian formulation

If we want true Brownian forcing, a standard mathematical description is to use white noise as the formal derivative of Brownian motion:

$$
\boxed{
\xi(t)=\frac{dW_t}{dt}
}
$$

in a formal sense, where $W_t$ is standard Brownian motion.

So the stochastic beam model can be written as

$$
\boxed{
y''+\omega_1^2 y=\frac{\sigma}{\mu}\,\dot W_t
}
$$

or equivalently in differential form,

$$
\boxed{
d v = -\omega_1^2 y\,dt + \frac{\sigma}{\mu}\,dW_t,
\qquad
dy = v\,dt,
}
$$

where $v=y'$.

This captures random kicks in time, which is a reasonable model for Brownian-type fluctuations.
"""))
```

```python
display(Markdown(r"""
## Problem 3

Now I can numerically simulate a beam mode with Brownian forcing using the stochastic system

$$
dy=v\,dt,
\qquad
dv=-\omega_1^2 y\,dt+\frac{\sigma}{\mu}\,dW_t.
$$

We use Euler-Maruyama on the first mode.
"""))

display(Markdown(r"""
### References

Shen, G., Wang, J., & Zhang, X. (2020).
*Euler–Maruyama method for distribution dependent stochastic differential equation driven by multiplicative fractional Brownian motion.*
"""))
```

```python
# Parameters
L = 1.0
E = 1.0
I = 1.0
mu = 1.0
sigma = 0.6

omega1 = (np.pi / L)**2 * np.sqrt(E * I / mu)

# Time discretization
T = 2.0
dt = 0.001
N = int(T / dt)
t = np.linspace(0, T, N + 1)

# Brownian increments
dW = np.sqrt(dt) * np.random.randn(N)

# Arrays
y = np.zeros(N + 1)
v = np.zeros(N + 1)

# Euler-Maruyama (copied from online literature like we were allowed to)
for n in range(N):
    y[n+1] = y[n] + v[n] * dt
    v[n+1] = v[n] - omega1**2 * y[n] * dt + (sigma / mu) * dW[n]
```

```python
plt.figure(figsize=(8,5))
plt.plot(t, y)
plt.xlabel(r"$t$")
plt.ylabel(r"Modal amplitude $y(t)$")
plt.title("First-Mode Response with Brownian Forcing")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, L, 400)
snapshot_indices = [0, int(0.5/dt), int(1.0/dt), int(1.5/dt), int(2.0/dt)]

plt.figure(figsize=(9,6))
for idx in snapshot_indices:
    w_snapshot = y[idx] * np.sin(np.pi * x / L)
    plt.plot(x, w_snapshot, label=rf"$t={t[idx]:.2f}$")

plt.xlabel(r"$x$")
plt.ylabel(r"$w(x,t)$")
plt.title("Beam Shape Snapshots Under Brownian Forcing")
plt.grid(True)
plt.legend()
plt.show()
```

```python
display(Markdown(r"""
## Verifying the stochastic model

A reasonable argument is based on the fact that Brownian increments satisfy

$$
\mathbb{E}[dW_n]=0,
\qquad
\mathrm{Var}(dW_n)=dt.
$$

Therefore, in the numerical method:
- the random increments should have mean close to $0$,
- their variance should be close to $dt$,
- increasing $\sigma$ should increase the typical size of the response,
- setting $\sigma=0$ should (at least I hope) recover the unforced oscillator.

Below we verify the statistics numerically.
"""))
```
