### Problem 1

```python
# Problem 1
print("Deriving the General Spring–Mass Model")
print("(a) Forces Acting on the Mass\n")

print("1. Spring force:")
print("   The spring exerts a force that tries to return the mass to its equilibrium position.")
print("   By Hooke’s Law, the force is proportional to displacement and acts in the opposite direction:")
print("   F_spring = -k x")
print("   where k > 0 is the spring constant and x(t) is the displacement.\n")

print("2. Damping force:")
print("   Damping represents resistive effects such as friction.")
print("   This force is proportional to velocity and acts opposite to the direction of motion:")
print("   F_damping = -c x'")
print("   where c > 0 is the damping coefficient and x'(t) is the velocity.\n")

print("3. External force:")
print("   An external force may act on the mass due to a driving mechanism.")
print("   This force is typically time-dependent:")
print("   F_external = F(t)")
print("   where F(t) is a prescribed function of time.\n")

print("4. Gravitational force:")
print("   Gravity exerts a constant force mg on the mass.")
print("   In many spring–mass models, gravity is absorbed into the equilibrium position")
print("   and does not appear explicitly in the equation of motion.\n")

print("KEY IDEA:")
print("The total force acting on the mass is the sum of the spring force, damping force,")
print("and any external force. Applying Newton’s Second Law you get:")
print("m*x'' + c*x' + k*x = F(t)")
```

### Problem 2

```python
# Problem 2
print("Problem 2: Undamped and Unforced Spring–Mass System\n")

print("Assumptions:")
print("- No friction (c = 0)")
print("- No external force (F(t) = 0)\n")

print("The governing equation is:")
print("m x'' + k x = 0\n")

print("This is a second-order linear ODE describing simple harmonic motion.")
print("Its general solution is:")
print("x(t) = A cos(ω t) + B sin(ω t)")
print("where ω = sqrt(k / m)\n")

print("The constants m and k determine the angular frequency ω.")
print("- Increasing k increases ω (faster oscillations)")
print("- Increasing m decreases ω (slower oscillations)")
print("- Neither m nor k affects the amplitude unless initial conditions change\n")
```

```python
# Varying the spring constant

# Initial conditions
x0 = 0.1      # initial displacement
v0 = 0.0      # initial velocity

# Time array
t = np.linspace(0, 10, 2000)

# Fixed mass
m = 1.0

# Different spring constants
k_values = [4, 9, 16]

for k in k_values:
    omega = np.sqrt(k / m)
    x = x0 * np.cos(omega * t) + (v0 / omega) * np.sin(omega * t)
    plt.plot(t, x, label=f"k = {k}")

plt.xlabel("Time t")
plt.ylabel("Displacement x(t)")
plt.title("Undamped for Different Spring Constants")
plt.legend()
plt.grid(True)
plt.show()
plt.figure()

print("Figure interpretation:")
print("- All solutions oscillate sinusoidally with constant amplitude.")
print("- Larger values of k result in higher frequency oscillations.")
print("- The amplitude remains constant because there is no damping.\n")
```

```python
# Varying the mass

# Different masses
m_values = [0.5, 1.0, 2.0]
k = 9.0

for m in m_values:
    omega = np.sqrt(k / m)
    x = x0 * np.cos(omega * t)
    plt.plot(t, x, label=f"m = {m}")

plt.xlabel("Time t")
plt.ylabel("Displacement x(t)")
plt.title("Undamped for Different Masses")
plt.legend()
plt.grid(True)
plt.show()
plt.figure()

print("Additional notes that I want to highlight:")
print("- Increasing the mass decreases the oscillation frequency.")
print("- Heavier masses oscillate more slowly.")
print("- Amplitude is unchanged when initial displacement is fixed.")
```

### Problem 3

```python
print("Problem: Damped, Unforced Spring–Mass System\n")
print("Assumptions I am making:")
print("- Friction exists (c > 0)")
print("- No external forcing (F(t) = 0)\n")

print("Model equation:")
print("m x'' + c x' + k x = 0\n")

print("KEY IDEA: behavior depends on the discriminant c^2 - 4mk.")
print("- Underdamped: c^2 < 4mk  -> oscillations with decaying amplitude")
print("- Critically damped: c^2 = 4mk -> fastest return to equilibrium with no oscillation")
print("- Overdamped: c^2 > 4mk  -> no oscillation, slower return (sum of decaying exponentials)\n")

# Exact solution formulas
def x_underdamped(t, m, c, k, x0, v0):
    omega0 = np.sqrt(k/m)
    alpha  = c/(2*m)
    omega_d = np.sqrt(omega0**2 - alpha**2)
    # x(t) = e^{-alpha t}[ C1 cos(omega_d t) + C2 sin(omega_d t) ]
    C1 = x0
    C2 = (v0 + alpha*x0)/omega_d
    return np.exp(-alpha*t) * (C1*np.cos(omega_d*t) + C2*np.sin(omega_d*t))

def x_critical(t, m, c, k, x0, v0):
    alpha = c/(2*m)  # here it is good to point out alpha^2 = k/m
    # x(t) = (C1 + C2 t) e^{-alpha t}
    C1 = x0
    C2 = v0 + alpha*x0
    return (C1 + C2*t) * np.exp(-alpha*t)

def x_overdamped(t, m, c, k, x0, v0):
    # roots r1, r2 real and negative
    disc = c**2 - 4*m*k
    r1 = (-c + np.sqrt(disc))/(2*m)
    r2 = (-c - np.sqrt(disc))/(2*m)
    # Solve for constants in x(t)=A e^{r1 t} + B e^{r2 t}
    # Using x(0)=x0 => A+B=x0
    # v(0)=v0 => A r1 + B r2 = v0
    A = (v0 - r2*x0)/(r1 - r2)
    B = x0 - A
    return A*np.exp(r1*t) + B*np.exp(r2*t)


# We can setup a loop "helper" so to speak for the solution
def classify_regime(m, c, k):
    D = c**2 - 4*m*k
    if np.isclose(D, 0.0):
        return "critical"
    elif D < 0:
        return "underdamped"
    else:
        return "overdamped"

def x_solution(t, m, c, k, x0, v0):
    reg = classify_regime(m, c, k)
    if reg == "underdamped":
        return x_underdamped(t, m, c, k, x0, v0)
    elif reg == "critical":
        return x_critical(t, m, c, k, x0, v0)
    else:
        return x_overdamped(t, m, c, k, x0, v0)

# Parameter study
x0, v0 = 0.1, 0.0
t = np.linspace(0, 10, 3000)

m = 1.0
k = 16.0

# Critical damping value: c_crit = 2*sqrt(mk)
c_crit = 2*np.sqrt(m*k)

print(f"Using the parameters m={m}, k={k}.")
print(f"Critical damping threshold: c_crit = 2*sqrt(mk) = {c_crit:.4f}\n")

# Choose c values around critical
c_values = [0.25*c_crit, 1.0*c_crit, 2.5*c_crit]
labels = ["underdamped (c < c_crit)", "critical (c = c_crit)", "overdamped (c > c_crit)"]

for c, lab in zip(c_values, labels):
    x = x_solution(t, m, c, k, x0, v0)
    plt.plot(t, x, label=f"{lab}, c={c:.3f}")

plt.xlabel("Time t")
plt.ylabel("Displacement x(t)")
plt.title("Damped, Unforced Spring–Mass System")
plt.legend()
plt.grid(True)
plt.show()
plt.figure()

print("Figure interpretation of the cases:")
print("- Underdamped: oscillates but amplitude decays exponentially (energy loss to friction).")
print("- Critical: returns to equilibrium fastest without oscillating.")
print("- Overdamped: returns without oscillating but slower than critical.\n")

print("How does the constants influence behavior:")
print("- Larger c (more friction) increases damping and can remove excess oscillations that you may see.")
print("- Larger k (stiffer spring) increases natural frequency and raises c_crit = 2*sqrt(mk).")
print("- Larger m (heavier mass) lowers natural frequency and also raises c_crit (since sqrt(mk) increases).\n")
```

```python
import sympy as sp
# Symbols
t, s = sp.symbols('t s', real=True)

# Part (c)
# X(s) = [ (s+2)/10 + 3/(s^2+9) ] / (s^2 + 2s + 16)
Xs = ((s + 2)/10 + 3/(s**2 + 9)) / (s**2 + 2*s + 16)

x_t = sp.simplify(sp.inverse_laplace_transform(Xs, s, t))

print("x(t) =")
sp.pprint(x_t)

# Part (d) steady-state and transient
# x_ss(t) = (7/85) sin(3t) - (6/85) cos(3t)
# x_tr(t) = x(t) - x_ss(t)

x_ss = (7/85)*sp.sin(3*t) - (6/85)*sp.cos(3*t)
x_tr = sp.simplify(x_t - x_ss)

print("\nSteady-state x_ss(t) =")
sp.pprint(x_ss)

print("\nTransient x_tr(t) =")
sp.pprint(x_tr)

print("\nDamping explanation:")
print("Because c>0, the transient contains an exponential decay factor e^{-(c/2m)t} = e^{-t},")
print("so the transient oscillations die out and the solution approaches the steady-state oscillation.")

# Plot x(t), x_ss(t), x_tr(t)

x_num   = sp.lambdify(t, x_t,  "numpy")
xss_num = sp.lambdify(t, x_ss, "numpy")
xtr_num = sp.lambdify(t, x_tr, "numpy")

tt = np.linspace(0, 15, 2000)

plt.figure()
plt.plot(tt, x_num(tt), label="x(t)")
plt.plot(tt, xss_num(tt), label="x_ss(t)")
plt.plot(tt, xtr_num(tt), label="x_tr(t)")
plt.xlabel("t (s)")
plt.ylabel("x (m)")
plt.title("Mass–Spring–Damper Response")
plt.grid(True)
plt.legend()
plt.show()

# Part (e) damping type using c^2 vs 4mk
m, c, k = 1, 2, 16
if c**2 < 4*m*k:
    damping_type = "underdamped"
elif c**2 == 4*m*k:
    damping_type = "critically damped"
else:
    damping_type = "overdamped"

print(f"\nDamping classification:")
print(f"c^2 = {c**2}, 4mk = {4*m*k} -> system is {damping_type}.")
```

```python
from scipy.integrate import solve_ivp
```

```python
# Numbers from the problem
m1, m2, m3 = 0.1, 0.15, 0.3
k1, k2, k3 = 20, 35, 40

M = np.array([[m1, 0,  0],
              [0,  m2, 0],
              [0,  0,  m3]], dtype=float)

K = np.array([[k1+k2, -k2,   0],
              [-k2,   k2+k3, -k3],
              [0,     -k3,   k3]], dtype=float)

# Part (b)
# A v = (omega^2) v

A = np.linalg.inv(M) @ K
lam, V = np.linalg.eig(A)   # lam = omega^2

lam = lam.real
V = V.real

# sort smallest to largest
idx = np.argsort(lam)
lam = lam[idx]
V = V[:, idx]

omega = np.sqrt(lam)

print("omega^2 =", lam)
print("omega (rad/s) =", omega)

# normalize mode shapes by max abs entry
Vn = V.copy()
for j in range(3):
    Vn[:, j] = Vn[:, j] / np.max(np.abs(Vn[:, j]))

print("\nMode shapes (columns, normalized):")
print(Vn)

# I plot mode shapes now
mass_index = [1, 2, 3]
plt.figure()
for j in range(3):
    plt.plot(mass_index, Vn[:, j], marker="o", label=f"Mode {j+1}")
plt.axhline(0)
plt.xticks([1,2,3], ["m1","m2","m3"])
plt.xlabel("Mass")
plt.ylabel("Relative displacement")
plt.title("Mode shapes (normalized)")
plt.grid(True)
plt.legend()
plt.show()

# Part (c): time evolution using ode45-style solver
# y = [x1,x2,x3,v1,v2,v3]
# x' = v
# v' = -M^{-1}K x
Minv = np.linalg.inv(M)

def rhs(t, y):
    x = y[0:3]
    v = y[3:6]
    xdot = v
    vdot = -(Minv @ (K @ x))
    return np.concatenate([xdot, vdot])

# initial condition: x3=0.01, others 0, released from rest
y0 = np.array([0.0, 0.0, 0.01,  0.0, 0.0, 0.0])

t_span = (0, 20)
t_eval = np.linspace(0, 20, 4000)

sol = solve_ivp(rhs, t_span, y0, method="RK45", t_eval=t_eval)

t = sol.t
x1 = sol.y[0]
x2 = sol.y[1]
x3 = sol.y[2]

plt.figure()
plt.plot(t, x1, label="x1(t)")
plt.plot(t, x2, label="x2(t)")
plt.plot(t, x3, label="x3(t)")
plt.xlabel("t (s)")
plt.ylabel("Displacement (m)")
plt.title("x1(t), x2(t), x3(t) with x3(0)=0.01 and rest released")
plt.grid(True)
plt.legend()
plt.show()

print("""My thoughts on the theory:
I interpret this by assuming sinusoidal motion. The equations reduce to an eigenvalue problem whose solutions give the
natural frequencies and mode shapes of the system. Each eigenvalue corresponds to a squared natural
frequency, and each eigenvector describes a normal mode in which all three masses oscillate at the same
frequency with fixed relative amplitudes. The lowest-frequency mode represents a motion
where the masses move in the same direction, while higher-frequency modes involve opposing motion
between adjacent masses, leading to greater spring deformation. Because the system is undamped, any
initial displacement can be expressed as a linear combination of these normal modes. When only the bottom
mass is initially displaced, multiple modes are excited, causing energy to transfer back and forth
between the masses without decay.
""")
```
