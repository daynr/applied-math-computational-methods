```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown
```

```python
def md(s):
    display(Markdown(s))

Nx = 201
Ny = 201
x = np.linspace(0, 1, Nx)
y = np.linspace(0, 1, Ny)
X, Y = np.meshgrid(x, y)

# Choose 6 time points to fill
tlist = np.array([0.0, 0.005, 0.01, 0.02, 0.05, 0.10])

lam = 2 * np.pi**2

# Problem 1 (Dirichlet, analytical solution)

md(f"""
## Problem 1

We solve on $\\Omega=(0,1)\\times(0,1)$ with $\\kappa=1$:

$$
u_t = u_{{xx}} + u_{{yy}}, \qquad u|_{{\\partial\\Omega}}=0.
$$

Initial condition:

$$
u(x,y,0)=\\sin(\\pi x)\\sin(\\pi y).
$$

Because $\\sin(\\pi x)\\sin(\\pi y)$ is a Laplacian eigenfunction,

$$
\\Delta(\\sin(\\pi x)\\sin(\\pi y))
= -2\\pi^2 \\sin(\\pi x)\\sin(\\pi y).
$$

So the solution is

$$
u(x,y,t)=e^{{-2\\pi^2 t}}\\sin(\\pi x)\\sin(\\pi y).
$$

**Long-time behavior:**

$$
u(x,y,t) \\to 0 \quad \\text{{as}} \quad t \\to \\infty.
$$
""")

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.ravel()

for k, t in enumerate(tlist):
    U = np.exp(-lam * t) * np.sin(np.pi * X) * np.sin(np.pi * Y)  # Solution

    ax = axes[k]
    im = ax.imshow(U, origin="lower", extent=[0, 1, 0, 1], aspect="equal")
    ax.set_title(f"Dirichlet: t = {t:g}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()
```

```python
# Problem 2

md(f"""
## Problem 2

We solve

$$
u_t = u_{{xx}} + u_{{yy}}, \qquad \\frac{{\\partial u}}{{\\partial n}}\\Big|_{{\\partial\\Omega}}=0.
$$

Initial condition:

$$
u(x,y,0)=1+\\cos(\\pi x)\\cos(\\pi y).
$$

The constant mode $1$ has eigenvalue $0$ (it does not decay).

The cosine mode satisfies

$$
\\Delta(\\cos(\\pi x)\\cos(\\pi y))
= -2\\pi^2 \\cos(\\pi x)\\cos(\\pi y).
$$

So the solution is

$$
u(x,y,t)=1+e^{{-2\\pi^2 t}}\\cos(\\pi x)\\cos(\\pi y).
$$

**Long-time behavior:**

$$
u(x,y,t) \\to 1 \quad \\text{{as}} \quad t \\to \\infty.
$$

### Mass Conservation

For Neumann (no-flux) boundaries, the total mass

$$
M(t)=\\int_\\Omega u(x,y,t) \\, dx \\, dy
$$

is conserved.

Numerically we approximate

$$
M(t) \\approx \\sum_{{i,j}} U_{{i,j}}(t) \\, \\Delta x \\, \\Delta y.
$$

Initial mass:

$$
M(0) \\approx {M[0]:.8f}
$$

Final mass:

$$
M(t_{{\\max}}) \\approx {M[-1]:.8f}
$$
""")

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.ravel()

for k, t in enumerate(tlist):
    U = 1.0 + np.exp(-lam * t) * np.cos(np.pi * X) * np.cos(np.pi * Y)

    ax = axes[k]
    im = ax.imshow(U, origin="lower", extent=[0, 1, 0, 1], aspect="equal")
    ax.set_title(f"Neumann: t = {t:g}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()

# Checking the mass conservation property we talked about in class
dx = x[1] - x[0]
dy = y[1] - y[0]

M = np.zeros_like(tlist, dtype=float)
for k, t in enumerate(tlist):
    U = 1.0 + np.exp(-lam * t) * np.cos(np.pi * X) * np.cos(np.pi * Y)
    M[k] = np.sum(U) * dx * dy

md(r"""
### Mass conservation plot
If Neumann BCs imply no flux across $\partial\Omega$, then $M(t)$ should be constant.
""")

plt.figure(figsize=(6,4))
plt.plot(tlist, M, marker="o")
plt.xlabel("t")
plt.ylabel("Discrete mass $M(t)$")
plt.title("Problem 2: Mass vs time (should be constant)")
plt.grid(True)
plt.show()

md(f"""
$$
M(0)\\approx {M[0]:.8f}, \\qquad M(t_{{\\max}})\\approx {M[-1]:.8f}.
$$

Since this is the analytical solution evaluated on the grid, the mass is constant up to machine precision.
""")
```

```python
# Problem 3

md(r"""
## Problem 3

We solve on $\Omega=(0,1)\times(0,1)$ with $\kappa=1$:
$$
u_t = u_{xx}+u_{yy}.
$$

Boundary conditions:
$$
u(0,y,t)=1,\quad u(1,y,t)=1,\quad u(x,0,t)=0,\quad u(x,1,t)=0.
$$

Initial condition:
$$
x_0=\tfrac12,\; y_0=\tfrac12,\; \sigma=0.1,\qquad
u(x,y,0)=\exp\!\left(-\frac{(x-x_0)^2+(y-y_0)^2}{2\sigma^2}\right).
$$

We will:
1) compute the transient solution numerically and plot heat maps at multiple times;
2) compute a numerical steady state by solving $\Delta u_\infty=0$;
3) compute an analytical steady state (Fourier series, truncated);
4) compare numerical vs analytical steady states.
""")

kappa = 1.0
uL, uR = 1.0, 1.0

Nx, Ny = 161, 161
x = np.linspace(0, 1, Nx)
y = np.linspace(0, 1, Ny)
dx = x[1] - x[0]
dy = y[1] - y[0]
X, Y = np.meshgrid(x, y)

# Initial condition (Gaussian)
x0, y0, sigma = 0.5, 0.5, 0.1
u = np.exp(-(((X-x0)**2 + (Y-y0)**2)/(2*sigma**2)))

# Impose BCs on IC
u[:, 0]  = uL
u[:, -1] = uR
u[0, :]  = 0.0
u[-1, :] = 0.0

# Time step
# dt <= 1/(2*kappa*(1/dx^2 + 1/dy^2))
dt_stable = 1.0 / (2.0*kappa*(1.0/dx**2 + 1.0/dy**2))
dt = 0.4 * dt_stable

tlist = np.array([0.0, 0.005, 0.01, 0.02, 0.05, 0.10])  # 6 snapshots
tFinal = float(tlist[-1])
nSteps = int(np.ceil(tFinal/dt))
dt = tFinal/nSteps

snap_steps = np.array([int(round(t/dt)) for t in tlist], dtype=int)
snap_steps = np.clip(snap_steps, 0, nSteps)
snap_set = set(snap_steps.tolist())

snaps = {0: u.copy()}


for n in range(1, nSteps+1):
    unew = u.copy()
    unew[1:-1, 1:-1] = u[1:-1, 1:-1] + dt*kappa*(
        (u[1:-1, 2:] - 2*u[1:-1, 1:-1] + u[1:-1, :-2]) / dx**2 +
        (u[2:, 1:-1] - 2*u[1:-1, 1:-1] + u[:-2, 1:-1]) / dy**2
    )

    # Re-impose BCs
    unew[:, 0]  = uL
    unew[:, -1] = uR
    unew[0, :]  = 0.0
    unew[-1, :] = 0.0

    u = unew

    if n in snap_set:
        snaps[n] = u.copy()

u_transient_final = u.copy()

md(r"""
### Transient heat maps
We show heat maps at 6 times.
""")

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.ravel()

def plot_heat(ax, U, title):
    im = ax.imshow(U, origin="lower", extent=[0,1,0,1], aspect="equal")
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

for k, t in enumerate(tlist):
    n = int(round(t/dt))
    plot_heat(axes[k], snaps[n], f"t = {t:g}")

plt.tight_layout()
plt.show()

md(r"""
### Long-time expectation
As $t\to\infty$, $u(x,y,t)$ approaches a steady state $u_\infty$ satisfying:
$$
\Delta u_\infty = 0
$$
with the same boundary conditions.
""")


md(r"""
## Numerical steady state $u_\infty^{num}$

We solve the steady state problem:
$$
\Delta u_\infty = 0
$$
with the same Dirichlet boundary conditions using Jacobi iteration.
""")

u_ss_num = np.zeros_like(u_transient_final)
u_ss_num[:, 0]  = uL
u_ss_num[:, -1] = uR
u_ss_num[0, :]  = 0.0
u_ss_num[-1, :] = 0.0

tol = 1e-6
max_iter = 20000

for it in range(max_iter):
    u_old = u_ss_num.copy()

    u_ss_num[1:-1, 1:-1] = (
        (u_old[1:-1, 2:] + u_old[1:-1, :-2]) * dy**2 +
        (u_old[2:, 1:-1] + u_old[:-2, 1:-1]) * dx**2
    ) / (2*(dx**2 + dy**2))

    # BCs
    u_ss_num[:, 0]  = uL
    u_ss_num[:, -1] = uR
    u_ss_num[0, :]  = 0.0
    u_ss_num[-1, :] = 0.0

    err = np.max(np.abs(u_ss_num - u_old))
    if err < tol:
        break

md(f"""
Jacobi iterations: **{it+1}**
Final update error:
$$\\|u^{{(k+1)}}-u^{{(k)}}\\|_\\infty \\approx {err:.2e}.$$
""")


md(r"""
## Analytical steady state $u_\infty^{ana}$ (Fourier series)

We solve
$$
\Delta u_\infty = 0,\quad
u_\infty(0,y)=1,\;u_\infty(1,y)=1,\;u_\infty(x,0)=0,\;u_\infty(x,1)=0.
$$

Using separation of variables with sine modes in $y$, the series can be written as:
$$
u_\infty(x,y)=\sum_{k=1}^{\infty}\frac{4}{(2k-1)\pi}
\left(
\frac{\sinh((2k-1)\pi x)}{\sinh((2k-1)\pi)}
+
\frac{\sinh((2k-1)\pi(1-x))}{\sinh((2k-1)\pi)}
\right)\sin((2k-1)\pi y).
$$

To avoid numerical overflow from $\sinh((2k-1)\pi)$ for large $k$, we evaluate the ratios
$\sinh(a z)/\sinh(a)$ using a stable exponential formula.
""")

def sinh_ratio_stable(a, z):
    """
    Uses exponentials to avoid overflow in sinh(a).
    """
    num = np.exp(a*(z - 1.0)) - np.exp(-a*(z + 1.0))
    den = 1.0 - np.exp(-2.0*a)
    return num / den

def u_steady_analytic_stable(X, Y, Nodd=120):
    """
    Truncated odd-mode series (m=1,3,5,...,2*Nodd-1) with sinh ratios.
    """
    uA = np.zeros_like(X, dtype=float)
    for k in range(1, Nodd+1):
        m = 2*k - 1
        a = m*np.pi
        coef = 4.0/(m*np.pi)
        right = sinh_ratio_stable(a, X)        # sinh(a*x)/sinh(a)
        left  = sinh_ratio_stable(a, 1.0 - X)  # sinh(a*(1-x))/sinh(a)
        uA += coef * (right + left) * np.sin(a*Y)
    return uA

# Choose truncation cutoff
Nodd = 120
u_ss_ana = u_steady_analytic_stable(X, Y, Nodd=Nodd)

md(f"""
We truncate to **{Nodd}** odd terms (modes $1,3,5,\\dots,{2*Nodd-1}$).
""")


abs_err = np.abs(u_ss_num - u_ss_ana)
max_err = np.nanmax(abs_err)
l2_err = np.sqrt(np.nansum(abs_err**2) * dx * dy)

md(f"""
## Comparison: numerical vs analytical steady state

$$
\\|u_\\infty^{{num}}-u_\\infty^{{ana}}\\|_\\infty \\approx {max_err:.3e},
\\qquad
\\|u_\\infty^{{num}}-u_\\infty^{{ana}}\\|_2 \\approx {l2_err:.3e}.
$$

Largest errors typically appear near corners due to the sharp change in boundary data and series truncation.
""")

def show_heat_robust(U, title):
    Uclean = U[np.isfinite(U)]
    if Uclean.size == 0:
        print("No finite values to plot.")
        return
    vmin, vmax = np.percentile(Uclean, [1, 99])
    plt.figure(figsize=(5.8,4.8))
    plt.imshow(U, origin="lower", extent=[0,1,0,1], aspect="equal", vmin=vmin, vmax=vmax)
    plt.colorbar()
    plt.title(title)
    plt.xlabel("x"); plt.ylabel("y")
    plt.show()

show_heat_robust(u_ss_num, "Numerical steady state $u_\\infty^{num}$")
show_heat_robust(u_ss_ana, f"Analytical steady state $u_\\infty^{{ana}}$ (stable series, {Nodd} odd terms)")
show_heat_robust(abs_err,  "Error heatmap $|u_\\infty^{num}-u_\\infty^{ana}|$")
```

```python
# Problem 4

md(r"""
## Problem 4

Robin BC (with $u_\infty=0$):
$$
-\frac{\partial u}{\partial n} = h\,u \quad \text{on } \partial\Omega.
$$

### (a) For $h=0$, what does the problem reduce to?
If $h=0$ then
$$
-\frac{\partial u}{\partial n}=0 \iff \frac{\partial u}{\partial n}=0,
$$
so it reduces to **homogeneous Neumann (insulated / no-flux)** boundary conditions.

### (b) For sufficiently large $h$, what does the problem “reduce” to?
If $h\gg 1$, the condition $-\partial_n u = h u$ strongly penalizes nonzero boundary values,
so effectively
$$
u|_{\partial\Omega}\approx 0,
$$
meaning it behaves like **homogeneous Dirichlet** boundary conditions.

### (c) Physical argument: what happens to energy for $h>0$?
Robin BC models **convective heat loss** to ambient air. When $h>0$, heat flows out through
the boundary at a rate proportional to $u$, so the system’s energy decreases.

A standard “energy” is the $L^2$ energy:
$$
E(t)=\frac12\int_\Omega u(x,y,t)^2\,dx\,dy.
$$
Using integration by parts and the Robin BC, one gets
$$
\frac{dE}{dt} = -\int_\Omega |\nabla u|^2\,dx\,dy \;-\; h\int_{\partial\Omega} u^2\,ds \;\le\; 0.
$$
So for $h>0$, energy decays.
""")

Nx = 161
Ny = 161
x = np.linspace(0, 1, Nx)
y = np.linspace(0, 1, Ny)
dx = x[1] - x[0]
dy = y[1] - y[0]
X, Y = np.meshgrid(x, y)  # shape (Ny, Nx)

# Gaussian bump IC
x0, y0, sigma = 0.5, 0.5, 0.1
u0 = np.exp(-(((X-x0)**2 + (Y-y0)**2) / (2*sigma**2)))

# Stability
dt_stable = 1.0 / (2.0 * (1.0/dx**2 + 1.0/dy**2))
dt = 0.4 * dt_stable
tFinal = 0.20
nSteps = int(np.ceil(tFinal/dt))
dt = tFinal / nSteps  # hit tFinal exactly

# Times to plot
tPlot = np.array([0.0, 0.01, 0.03, 0.06, 0.10, 0.20])
plotSteps = np.unique(np.clip(np.round(tPlot/dt).astype(int), 0, nSteps))

def apply_robin(U, h):
    if h == 0.0:
        U[:, 0]  = U[:, 1]
        U[:, -1] = U[:, -2]
        U[0, :]  = U[1, :]
        U[-1, :] = U[-2, :]
        return U

    U[:, 0]  = U[:, 1]  / (1.0 + h*dx)   # x=0
    U[:, -1] = U[:, -2] / (1.0 + h*dx)   # x=1
    U[0, :]  = U[1, :]  / (1.0 + h*dy)   # y=0
    U[-1, :] = U[-2, :] / (1.0 + h*dy)   # y=1
    return U

def l2_energy(U):
    return 0.5 * np.sum(U**2) * dx * dy

def run_robin(h, make_plots=True):
    u = u0.copy()
    u = apply_robin(u, h)  # enforce BC at t=0

    E = np.zeros(nSteps+1)
    E[0] = l2_energy(u)

    snaps = {}
    if 0 in plotSteps:
        snaps[0] = u.copy()

    for n in range(1, nSteps+1):
        # Laplacian on interior
        unew = u.copy()
        unew[1:-1, 1:-1] = u[1:-1, 1:-1] + dt * (
            (u[1:-1, 2:] - 2*u[1:-1, 1:-1] + u[1:-1, :-2]) / dx**2 +
            (u[2:, 1:-1] - 2*u[1:-1, 1:-1] + u[:-2, 1:-1]) / dy**2
        )

        # Apply Robin BC
        unew = apply_robin(unew, h)

        u = unew
        E[n] = l2_energy(u)

        if n in plotSteps:
            snaps[n] = u.copy()

    if make_plots:
        # 2x3 snapshot plot
        fig, axes = plt.subplots(2, 3, figsize=(14, 8))
        axes = axes.ravel()

        for k, t in enumerate(tPlot):
            n = int(np.round(t/dt))
            Uplot = snaps.get(n, u)
            im = axes[k].imshow(Uplot, origin="lower", extent=[0,1,0,1], aspect="equal")
            axes[k].set_title(f"h={h:g}, t={t:g}")
            axes[k].set_xlabel("x")
            axes[k].set_ylabel("y")
            plt.colorbar(im, ax=axes[k], fraction=0.046, pad=0.04)

        plt.tight_layout()
        plt.show()
        plt.close(fig)

        # energy plot
        tvec = np.linspace(0, tFinal, nSteps+1)
        plt.figure(figsize=(6,4))
        plt.plot(tvec, E, linewidth=2)
        plt.grid(True)
        plt.xlabel("t")
        plt.ylabel(r"$E(t)=\frac{1}{2}\int_{\Omega} u^2\,dx\,dy$")
        plt.title(f"L2 energy vs time (Robin BC), h={h:g}")
        plt.show()
        plt.close()

    return u, E

# Numerical verification for (a) and (b) and (c)

md(r"""
## Numerical verification

We run it now:
- $h=0$ (should behave like Neumann / insulated),
- moderate $h$ (e.g. $h=1$),
- large $h$ (e.g. $h=50$, should behave close to Dirichlet $u\approx 0$ on the boundary),

and we plot both snapshots and $L^2$ energy curves. For $h>0$, $E(t)$ should decrease faster.
""")

u_h0,  E_h0  = run_robin(h=0.0,  make_plots=True)
u_h1,  E_h1  = run_robin(h=1.0,  make_plots=True)
u_h50, E_h50 = run_robin(h=50.0, make_plots=True)

# Compare energy curves on one plot
tvec = np.linspace(0, tFinal, nSteps+1)
plt.figure(figsize=(7,4))
plt.plot(tvec, E_h0,  label="h=0 (Neumann)", linewidth=2)
plt.plot(tvec, E_h1,  label="h=1", linewidth=2)
plt.plot(tvec, E_h50, label="h=50 (~Dirichlet)", linewidth=2)
plt.grid(True)
plt.xlabel("t")
plt.ylabel(r"$E(t)$")
plt.title("Energy decay comparison for different h")
plt.legend()
plt.show()

md(r"""
### What you do observe
- **$h=0$:** boundary acts insulated (Neumann).
- **$h>0$:** energy decays **faster**, because the boundary removes heat (extra dissipation term $h\int_{\partial\Omega}u^2\,ds$).
- **Large $h$:** boundary values are forced close to $0$, so the behavior resembles Dirichlet cooling at the boundary.
""")
```
